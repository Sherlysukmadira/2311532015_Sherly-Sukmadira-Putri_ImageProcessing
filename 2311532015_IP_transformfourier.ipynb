```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data citra 3x3
citra = np.array([
    [148, 164, 174],
    [144, 159, 167],
    [128, 147, 157]
])

print("Citra asli (3x3):")
print(citra)
```

```python
# Melakukan DFT 2D menggunakan FFT dari NumPy
dft = np.fft.fft2(citra) #fungsi untuk melakukan Transformasi Fourier 2D

print("Hasil DFT (bilangan kompleks):")
print(dft)
```

```python
#mengubah hasil FFT yang kompleks menjadi nilai yang bisa dipahami/diamati.
magnitude = np.abs(dft)
print("Magnitude (spektrum):")
print(magnitude)
```

```python
# Ambil informasi fase (sudut) dari hasil Fourier Transform
phase = np.angle(dft)
print("Fase (radian):")
print(phase)
```

```python
# pisahkan dan tampilkan komponen penyusun utama dari hasil FFT
print("Bagian Real:")
print(dft.real)
print()
print("Bagian Imajiner:")
print(dft.imag)
```

```python
# Visualisasi
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# Citra asli
axes[0].imshow(citra, cmap='gray', interpolation='nearest')
axes[0].set_title('Citra Asli')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
for i in range(3):
    for j in range(3):
        axes[0].text(j, i, str(citra[i, j]), ha='center', va='center', color='red', fontsize=10)


# Phase Radiant
im2 = axes[1].imshow(phase, cmap='gray', interpolation='nearest')
axes[1].set_title('Phase Radiant')
plt.colorbar(im2, ax=axes[2])

# Magnitude spectrum
im = axes[2].imshow(magnitude, cmap='gray', interpolation='nearest')
axes[2].set_title('Magnitude Spectrum')
plt.colorbar(im, ax=axes[1])

plt.tight_layout()
plt.show()
```
